Inspect the actual saved holdout metrics and one prediction. Bootstrap trains the model; this notebook does not retrain it.

```python
from pathlib import Path
import sys
ROOT = Path.cwd() if (Path.cwd() / 'src').is_dir() else Path.cwd().parent
sys.path.insert(0, str(ROOT))
import json
metrics = json.loads((ROOT / 'models/model_metrics.json').read_text())
metrics
```

```python
from src.config import Settings
from src.tools.sql_tool import SQLTool
from src.models.train import FraudModel
settings = Settings()
transaction = SQLTool(settings).transaction('TXN_0000002')
risk = FraudModel(settings.artifacts).score(transaction)
risk
```
