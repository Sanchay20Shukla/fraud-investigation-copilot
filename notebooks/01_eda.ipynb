# PaySim investigation data exploration
Read-only exploration of the imported dataset. Labels are available offline for research, never in investigator tools. Run bootstrap first.

```python
from pathlib import Path
import sys
ROOT = Path.cwd() if (Path.cwd() / 'src').is_dir() else Path.cwd().parent
sys.path.insert(0, str(ROOT))
import pandas as pd
from sqlalchemy import select, func
from src.config import Settings
from src.db import engine_for, transactions, labels
engine = engine_for(Settings().database_url, readonly=True)
with engine.connect() as conn:
    counts = pd.read_sql(select(transactions.c.transaction_type, func.count().label('count'), func.avg(transactions.c.amount).label('average_amount')).group_by(transactions.c.transaction_type), conn)
counts
```

```python
counts.plot.bar(x='transaction_type', y='count', title='Imported transactions by type')
```

## Interpretation
Amounts use simulation units. Imported history may be incomplete. Same-step transactions have unknown ordering. This notebook describes the loaded sample, not a production fraud population.
